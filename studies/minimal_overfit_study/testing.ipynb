```python
from e3nn.o3 import Irreps, spherical_harmonics
import torch
import numpy as np
```

```python
edge_vecs = torch.randn(5, 3)  # 5 random edge vectors
l_max = 2
irreps_sh = Irreps.spherical_harmonics(l_max)
Y = spherical_harmonics(irreps_sh, edge_vecs, normalize=True, normalization='component')
Y
```
